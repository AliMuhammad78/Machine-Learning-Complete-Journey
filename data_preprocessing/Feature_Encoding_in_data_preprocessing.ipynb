 # Feature Encoding 

Feature encoding is the process of converting the categorical data to numerical data , because it is easy for computers to handle the numerical data than categorical data . 

Following are some methods of data encoding 
1. Label encoding
2. Ordinal encoding
3. One-hot encoding
4. Binary encoding
5. Manual Encoding

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder, OneHotEncoder , OrdinalEncoder
```

```python
df = sns.load_dataset('tips')
df.head()
```

```python
df["time"].value_counts()
```

## Label Encoding 

```python
encoder = LabelEncoder()
df["Encoded-time"] = encoder.fit_transform(df["time"]) 
df.head()
```

```python
print(df["Encoded-time"].value_counts())
```

## Ordinal Encoding 

```python
df["day"].value_counts()
```

```python
ordinal_encoder = OrdinalEncoder(categories=[["Thur", "Fri", "Sat", "Sun"]])
ordinal_encoder.fit_transform(df[["day"]])
```

# 3. One hot encoding

```python
titanic = sns.load_dataset('titanic')
one_hot_endcoder = OneHotEncoder()
embarked_encoded = one_hot_endcoder.fit_transform(titanic[["embarked"]])
embarked_encoded.toarray()
embarked_encoded_df = pd.DataFrame(embarked_encoded.toarray(), columns=one_hot_endcoder.get_feature_names_out(["embarked"]))
embarked_encoded_df.head()
titanic = pd.concat([titanic, embarked_encoded_df], axis=1)
titanic.head()
```
