## Ridge Regression (L2 Regularization)

### 🔹 Why is Ridge Regression Needed?
In Linear Regression, when:
- There are many features
- Features are highly correlated (multicollinearity)
- Model starts overfitting (fits training data too well)

👉 The model becomes unstable and performs poorly on new data.

Ridge Regression solves this by adding a penalty on large coefficients.

 
### 🔹 Formula

Loss Function:
Loss = MSE + λ * Σ(bᵢ²)

 

### 🔹 Key Idea
- Adds penalty equal to the square of coefficients
- Shrinks coefficients towards zero
- But never makes them exactly zero

 

### 🔹 Effects
- Reduces overfitting
- Handles multicollinearity well
- Keeps all features in the model

 

### 🔹 Role of λ (Lambda)
- λ = 0 → Normal Linear Regression
- λ increases → coefficients shrink more
- High λ → simpler model but may underfit

 

### 🔹 When to Use Ridge
- When all features are important
- When features are correlated
- When you want to reduce model complexity but not remove features
 

### 🔹 Summary
Ridge Regression reduces overfitting by shrinking coefficients without removing any feature.

---

```python
import pandas as pd 
import numpy as np 
from sklearn.linear_model import LinearRegression

X = np.array([[1 , 1], [ 1, 2], [2 ,3], [4 ,6], [5 ,11]])
y=np.dot(X, np.array([1, 2])) + 3

le = LinearRegression()

le.fit(X , y)

print("Coefficients" , le.coef_)
print("Intercept" , le.intercept_)
```

```python
from sklearn.linear_model import Ridge
ridge = Ridge(alpha=1.0)

ridge.fit(X , y)
print("Coefficients" , ridge.coef_)
print("Intercept" , ridge.intercept_)
```

# Comparing Simple Linear Regression vs. Ridge Regression

```python
import seaborn as sns
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, mean_absolute_percentage_error
from sklearn.preprocessing import OneHotEncoder, StandardScaler, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np

# Load the data set
df = sns.load_dataset('diamonds')
```

```python
X = df.drop('price', axis=1)
y = df['price']

# numeric features
numeric_features = ['carat', 'depth', 'table', 'x', 'y', 'z']
# categorical features
categorical_features = ['cut', 'color', 'clarity']

# preprocess the data
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(), categorical_features)
    ]
)

# train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
```

```python
lr_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                              ('regressor', LinearRegression())])

# Ridge Regression Pipeline
ridge_pipeline = Pipeline(steps=[('preprocessor', preprocessor),
                                 ('regressor', Ridge(alpha=1.5))])
```

### Comparison of two models

```python
# Train and evaluate Linear Regression
lr_pipeline.fit(X_train, y_train)
lr_pred = lr_pipeline.predict(X_test)
lr_mse = mean_squared_error(y_test, lr_pred)
lr_r2 = r2_score(y_test, lr_pred)
lr_mae = mean_absolute_error(y_test, lr_pred)
lr_mape = mean_absolute_percentage_error(y_test, lr_pred)
lr_rmse = np.sqrt(lr_mse)

# Train and evaluate Ridge Regression
ridge_pipeline.fit(X_train, y_train)
ridge_pred = ridge_pipeline.predict(X_test)
ridge_mse = mean_squared_error(y_test, ridge_pred)
ridhe_r2 = r2_score(y_test, ridge_pred)
ridge_mae = mean_absolute_error(y_test, ridge_pred)
ridge_mape = mean_absolute_percentage_error(y_test, ridge_pred)
ridge_rmse = np.sqrt(ridge_mse)

print("Linear Regression MSE:", lr_mse)
print("Ridge Regression MSE:", ridge_mse)
print(f"------------------------")

print("Linear Regression R2:", lr_r2)
print("Ridge Regression R2:", ridhe_r2)
print(f"------------------------")
print("Linear Regression MAE:", lr_mae)
print("Ridge Regression MAE:", ridge_mae)
print(f"------------------------")
print("Linear Regression MAPE:", lr_mape)
print("Ridge Regression MAPE:", ridge_mape)
print(f"------------------------")
print("Linear Regression RMSE:", lr_rmse)
print("Ridge Regression RMSE:", ridge_rmse)
```

>There is a suble difference as it is multiple linear regression , you have to adjust the values of alpha according to your need .
