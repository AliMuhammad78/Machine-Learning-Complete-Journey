```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
```

## Simple Linear Regression 
* simple - input variable is one 
* linear - it gives model in the form of line 
* Regression - because the prediction is value

```python
from sklearn.linear_model import LinearRegression
```

```python
# loading the data
df = sns.load_dataset('tips')
df.head()
```

```python
# select the features and target variable
# remember that the features should be in a 2D array and the target variable should be in a 1D array because that's what the scikit-learn library expects
X = df[['total_bill']]
y = df['tip']

```

### Seeing what is the tip on the basis of number of people on the table (size)

```python
# creating the regression line plot 
sns.lmplot(x="size", y="tip", data =df)
```

### Seeing what is the tip on the basis of total bill 

```python
sns.lmplot(x="total_bill", y="tip", data =df)
```

### Load the model and fit the model 
#### fitting means training of model 

```python
model = LinearRegression()
model.fit(X,y)
```

### Predicting 

```python
model.predict([[24]])
```

## Multiple Linear Regression 

* Multiple- Use more than one variables to predict

```python
X1 = df[['total_bill', 'size']]
model.fit(X1,y)
model.predict([[24, 4]])
```

## Encoding the Variable 
> Always remeber that Regression only take numerical variable , feature
>
> You need to encode the variable into numerical value 


```python
df["smoker_num"] = df["smoker"].map({"Yes": 1, "No": 0})
df.head(10)
```

```python
X2 = df[['total_bill', 'size', 'smoker_num']]
model.fit(X2,y)
print("Tip from Smoker =" , model.predict([[24, 4, 1]]))
print("Tip from Non-Smoker =" ,model.predict([[24, 4, 0]]))
```

 > Non smokers are giving slightly more tip than Smoker persons 

## Regression using Decision Tree

```python
from sklearn.tree import DecisionTreeRegressor 
model = DecisionTreeRegressor()

model.fit(X,y)
model.predict([[24]])
```
