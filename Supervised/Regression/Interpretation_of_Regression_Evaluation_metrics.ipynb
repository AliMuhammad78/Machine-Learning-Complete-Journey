# First: Why do we even have multiple metrics?

Because each metric looks at **error in a different way**.

Think like this:

👉 You predicted **diamond prices**
👉 Now you want to judge *how wrong you were*

But “wrong” can be measured in different styles:

* Big mistakes matter more? → MSE / RMSE
* Equal importance to all errors? → MAE
* Relative (percentage) error? → MAPE
* Overall fit quality? → R²

---

#  1. Mean Absolute Error (MAE)

👉 **Average of absolute errors**

### 💡 Meaning:

“How much am I wrong on average?”

###  Example:

If MAE = **500**
→ On average, your prediction is off by **$500**

###  Real-life analogy:

You estimate prices of phones:

* Actual: 50,000
* Predicted: 49,500

Error = 500

👉 MAE = “average mistake in rupees”

###  Interpretation rule:

* Lower = better
* Easy to understand
* Good when you want **simple average error**

---

#  2. Mean Squared Error (MSE)

👉 Squares the errors before averaging

### 💡 Meaning:

“Punishes large mistakes heavily”

### 📌 Example:

Error = 100 → becomes 10,000
Error = 1000 → becomes 1,000,000 😨

👉 Big mistakes become VERY important

###  Real-life analogy:

Teacher checking exams:

* Small mistake → okay
* Big mistake → heavy penalty

###  Interpretation rule:

* Lower = better
* Sensitive to outliers
* Used more in training than interpretation

---

#  3. Root Mean Squared Error (RMSE)

 Square root of MSE

###  Meaning:

Same as MSE but **back to original unit**

###  Example:

If RMSE = **700**
→ Your typical error ≈ **$700**

###  Real-life analogy:

Like MAE but:

* Penalizes big mistakes more

###  Interpretation rule:

* Lower = better
* Compare with MAE:

  * RMSE >> MAE → model making big mistakes somewhere

---

#  4. Mean Absolute Percentage Error (MAPE)

 Error in percentage

###  Meaning:

“How much % wrong am I?”

###  Example:

MAPE = 0.10 → **10% error**

###  Real-life analogy:

You predict salary:

* Actual: 100,000
* Predicted: 90,000

Error = 10%

👉 Easy to explain to non-technical people

### ⚠️ Problem:

* Bad when actual value ≈ 0
* Can explode

###  Interpretation rule:

* Lower = better
* <10% → very good
* 10–20% → good
* > 30% → poor

---

#  5. R² Score (MOST IMPORTANT)

👉 Measures **how well model explains data**

###  Range:

* 1 → perfect model ✅
* 0 → useless model ❌
* <0 → worse than guessing 😬

###  Example:

R² = 0.90
→ Model explains **90% of variation**

###  Real-life analogy:

Imagine:

* Student marks depend on study hours

R² tells:
👉 “How much marks depend on study?”

---

#  Now: How to judge YOUR model (General Rules)

##  Step 1: Look at R² first

* > 0.8 → very good
* 0.6–0.8 → good
* 0.4–0.6 → okay
* <0.4 → weak

---

##  Step 2: Check MAE / RMSE

Compare with actual values:

Example:

* Average diamond price = 5000
* RMSE = 500

👉 Error = 10% → good

BUT:

* RMSE = 3000
  👉 Error huge → bad

---

##  Step 3: Compare RMSE vs MAE

* RMSE ≈ MAE → stable model ✅
* RMSE >> MAE → big errors exist ❌

---

##  Step 4: Use MAPE for intuition

* Easy to explain:

  * “Model is 12% wrong on average”

---

#  MOST IMPORTANT CONCEPT

👉 **There is NO single “best” metric**

You should:
✔ Use combination
✔ Interpret together

---

# 🟣 Final Decision Rule (VERY IMPORTANT)

Your model is GOOD if:

✔ High R²
✔ Low MAE
✔ RMSE not too large
✔ MAPE reasonably small

---

# 🧠 Final Real-Life Analogy (BEST ONE)

Imagine you are predicting house prices:

| Metric | Meaning                                       |
| ------ | --------------------------------------------- |
| MAE    | Average mistake (like average guessing error) |
| RMSE   | Punishes big wrong guesses                    |
| MAPE   | % error (easy to explain to client)           |
| R²     | Overall intelligence of model                 |

👉 A good model:

* Makes small mistakes (low MAE)
* Avoids huge mistakes (low RMSE)
* Has high understanding (high R²)

````python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder , StandardScaler , LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error , mean_absolute_percentage_error
from sklearn.model_selection import train_test_split

import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 

data = sns.load_dataset("diamonds")
X = data[["carat", "cut", "color", "clarity", "depth", "table", "x", "y", "z"]]
y = data["price"]

numeric_features = ["carat" , "depth" , "table" , "x" , "y" ,"z"]
categorical_features = ["cut" , "color" , "clarity"]

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_features),
        ('cat', OneHotEncoder(), categorical_features)
    ]
)

# train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)

# pipeline
pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        ('model', LinearRegression())
    ]
)

# fit the model
pipeline.fit(X_train, y_train)

# metric to evaluate the model
y_pred = pipeline.predict(X_test)

print(f"Mean Squared Error: {mean_squared_error(y_test, y_pred)}")
print(f"R2 Score: {r2_score(y_test, y_pred)}")
print(f"Mean Absolute Error: {mean_absolute_error(y_test, y_pred)}")
print(f"Mean Absolute Percentage Error: {mean_absolute_percentage_error(y_test, y_pred)}")
# root mean squared error
print(f"Root Mean Squared Error: {np.sqrt(mean_squared_error(y_test, y_pred))}")
````
