### Import Libraries

```python
import numpy as np 
import pandas as pd 
import seaborn as sns
from sklearn.linear_model import LogisticRegression 
```

### Loading the dataset using pandas 

### Adult Census Income Dataset: Data Dictionary

* **`age`**: The age of the person.
* **`workclass`**: The general sector the person works in (e.g., private company, government, self-employed).
* **`fnlwgt`**: A statistical weight representing how many people in the real world this specific row stands for.
* **`education`**: The highest degree or level of schooling completed.
* **`education-num`**: The exact same education level, just represented as a number (higher number = more education).
* **`marital-status`**: Whether the person is married, single, divorced, etc.
* **`occupation`**: The specific type of job the person does (e.g., sales, tech support, farming).
* **`relationship`**: The person's role in their household (e.g., husband, unmarried, own-child).
* **`race`**: The person's race or ethnicity.
* **`sex`**: The person's gender.
* **`capital-gain`**: Extra money made from investments (like selling stocks or property).
* **`capital-loss`**: Money lost from investments.
* **`hours-per-week`**: How many hours the person works in a typical week.
* **`native-country`**: The country where the person was born.
* **`income`**: The target variable showing if they make more (>50K) or less (<=50K) than $50,000 a year.

```python


# Define the column names based on the UCI documentation
columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num', 
    'marital-status', 'occupation', 'relationship', 'race', 'sex', 
    'capital-gain', 'capital-loss', 'hours-per-week', 'native-country', 'income'
]

# Load the dataset directly from the UCI web repository
df = pd.read_csv(
    "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data", 
    names=columns, 
    sep=r'\s*,\s*', # This cleans up the extra spaces around the text in the CSV
    engine='python'
)

print(df.head())
```

```python
print(type(df['income'][0]))
```

## Performing Logistic Regression 


### We will see on the basis of some features that the person will earn more than 50k or not. 

I am using these features for prediction : 
- best_features = [
    'age',
    'education-num',
    'capital-gain',
    'hours-per-week',
    'marital-status',
    'occupation'
]

```python
X = df[['age', 'education-num','capital-gain','hours-per-week','marital-status','occupation']]
y = df["income"]
#  This automatically finds the text columns and turns them into 1s and 0s
X_encoded = pd.get_dummies(X)
model = LogisticRegression()
model.fit(X_encoded, y) 
# predict the 20th row of the encoded data
prediction = model.predict(X_encoded.iloc[[20]])
print(prediction) 
```

## By Naive Bayes 

```python
from sklearn.naive_bayes import GaussianNB as GNB 

X = df[['age', 'education-num','capital-gain','hours-per-week','marital-status','occupation']]
y = df["income"]
#  This automatically finds the text columns and turns them into 1s and 0s
X_encoded = pd.get_dummies(X)
model = GNB()
model.fit(X_encoded, y) 
# predict the 20th row of the encoded data
prediction = model.predict(X_encoded.iloc[[20]])
print(prediction) 
```
